- Correr el código para obtener las reseñas de los 254 lugares que pertenecen a la red de ATMs del Banco República

```python
#Versión "Robusta" a cortes/interrupciones

import pandas as pd
import time
import json
from apify_client import ApifyClient

client = ApifyClient("mi_token_de_apify")

df_con_reviews = pd.read_csv("sucursales_filtradas_254.csv")

run_input = {
    "startUrls": [{"url": url} for url in df_con_reviews["url"]],
    "reviewsOrigin": "google",
    "maxReviews": 9999, #techo que abarca todas las reseñas (8365 en total); esta versión del actor no acepta "0" para que la cantidad de reseñas no tenga un tope
    "language": "es"
}

# Versión "robusta" a cortes/interrupciones
run = client.actor("compass/google-maps-reviews-scraper").start(run_input=run_input)
run_id = run["id"]
print("Run ID:", run_id)

with open("run_id_reviews.txt", "w") as f:
    f.write(run_id)

while True:
    try:
        status = client.run(run_id).get()["status"]
        print("Estado:", status)
        if status in ("SUCCEEDED", "FAILED", "ABORTED", "TIMED-OUT"):
            break
    except Exception as e:
        print(f"Error de conexión al chequear estado, reintentando: {e}")
    time.sleep(30)

dataset_id_reviews = client.run(run_id).get()["defaultDatasetId"]
print("Dataset listo:", dataset_id_reviews)

# Bajar los items y guardar el JSON crudo ANTES de aplanar a tabla
items = list(client.dataset(dataset_id_reviews).iterate_items())

with open("reviews_brou_raw.json", "w", encoding="utf-8") as f:
    json.dump(items, f, ensure_ascii=False, indent=2)

print(f"{len(items)} reviews guardadas en JSON crudo")
```

```python
# Se elimina la columna 'responseFromOwnerText', la información relevante ya queda capturada en la nueva variable
df = df.drop(columns=["responseFromOwnerText"])
df.info()
```

```python
# Nos interesan las reviews publicadas en el año 2025
# Primero ver cuántas reviews hay por año
df["publishedAtDate"].dt.year.value_counts().sort_index()
```

```python
# Filtramos por aquellas reviews publicadas en 2025
df = df[df["publishedAtDate"].dt.year == 2025].reset_index(drop=True)
print(f"Shape luego del filtro temporal: {df.shape}")
```

```python
# Normalizar 'text' para luego filtrar por keywords y quedarnos con las reviews que efectivamente sean de interés

import unicodedata
import re

def normalizar(texto):
    # 1. lowercase
    texto = texto.lower()
    # 2. sacar acentos
    texto = unicodedata.normalize('NFKD', texto)
    texto = ''.join(c for c in texto if not unicodedata.combining(c))
    # 3. eliminar caracteres especiales
    texto = re.sub(r'[^a-z0-9\s]', ' ', texto)
    # 4. colapsar espacios múltiples
    texto = re.sub(r'\s+', ' ', texto).strip()
    return texto

df["text_normalizado"] = df["text"].apply(normalizar)

# Verificación
print(df[["text", "text_normalizado"]].head(3))
```

```python
# Filtrar por keywords:

keywords = ["cajero", "automata", "buzonera", "reconocedora",
            "redbrou", "atm", "billete", "efectivo", "tarjeta", "maquina",
            "cajeros", "maquinas", "billetes"]

patron = "|".join(keywords)

mask = df["text_normalizado"].str.contains(patron, case=False, na=False)
print(f"Reviews que mencionan autómatas: {mask.sum()} de {len(df)}")

df_automatas = df[mask].reset_index(drop=True)
```

```python
df_automatas.info()
```

```python
# Creamos una nueva columna, "motivo_review", para indicar el tema que motivó la redacción de la review

motivos = {
    "dispensado": ["dispenso", "dispensar", "billete", "efectivo", "plata", "dinero"],
    "comprobante": ["comprobante", "ticket", "recibo", "imprimio", "imprime", "impresion"],
    "lentitud": ["lento", "lenta", "lentitud", "demora", "demoro", "tarda", "tardo", "espera", "cola", "fila"],
    "limpieza": ["sucio", "sucia", "limpieza", "mugre", "mugriento", "deteriorado"],
    "disponibilidad servicio": ["apagado", "apagada", "cerrado", "fuera de servicio", "no disponible", 
                                "sin servicio", "no funciona", "no anda", "inhabilitado", "falla", 
                                "fallo", "error", "trabo", "trabado", "trago", "atasca", "atascada", 
                                "no lee", "no reconoce", "bloqueado"]
}

def clasificar_motivo(texto):
    matches = []
    for motivo, keywords in motivos.items():
        patron = "|".join(keywords)
        if re.search(patron, texto):
            matches.append(motivo)
    return matches if matches else ["otro"]

df_automatas["motivo_review"] = df_automatas["text_normalizado"].apply(clasificar_motivo)

# Verificación
print(df_automatas[["text", "motivo_review"]].head(10))
print("\n--- Distribución de motivos ---")
from collections import Counter
todos_motivos = [m for lista in df_automatas["motivo_review"] for m in lista]
print(Counter(todos_motivos))
```

```python
# Para finalizar con el procesamiento del dataset, se crea una columna para indicar el "tono" de la review según la cantidad de estrellas otorgadas por el usuario
def clasificar_tono(stars):
    if stars <= 2:
        return "Negativo"
    elif stars == 3:
        return "Neutro"
    else:
        return "Positivo"

df_automatas["tono"] = df_automatas["stars"].apply(clasificar_tono)

print(df_automatas["tono"].value_counts())
```
